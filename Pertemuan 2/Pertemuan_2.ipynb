#### Pembagian kolom yang akan di encoder

```python
numerical_cols = ['tenure', 'MonthlyCharges', 'TotalCharges',
#     indeks         4             17              18
                  'SeniorCitizen']
#                      1
ordinal_cols = ['Contract']
# indeks            14
contract_categories = ['Month-to-month', 'One year', 'Two year']


categorical_cols = [
    'gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines',
    #  0         2            3             5                 6
    'InternetService', 'OnlineSecurity', 'OnlineBackup',
    #     7                   8                9
    'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies',
    #       10               11             12             13
    'PaperlessBilling', 'PaymentMethod'
    #       15                16
]
```

# Decision Tree Classifier

```python
# Pipeline
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression

# imputasi dan penskalaan untuk kolom numerical_col
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')), # mengisi data kosong memakai median
    ('scaler', StandardScaler()) # skala ulang kolom numerik
])

# imputasi dan penskalaan untuk kolom contract
ordinal_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder(categories=[contract_categories]))
])

# imputasi dan penskalaan untuk kolom categorial_cols
categorial_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ohe', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False))
])

prep = ColumnTransformer([
    ('numeric', numerical_transformer, numerical_cols),
    ('ordinal', ordinal_transformer, ordinal_cols),
    ('category', categorial_transformer, categorical_cols)
])

model2 = Pipeline([
    ('prep', prep),
    ('pilih', SelectKBest(f_classif, k=6)),
    ('model', DecisionTreeClassifier(max_depth=3,random_state=42))
])
```

```python
model2.fit(X_train, y_train)

score = model2.score(X_test, y_test)
print(f"Akurasi model pada data uji: {score:.4f}")
```

```python
train_sizes,train_scores, val_scores = learning_curve( model2,X_train,y_train,
    cv=5,train_sizes=np.linspace(0.1,1.0,10),
    scoring="accuracy" )
train_mean = train_scores.mean(axis=1)
val_mean = val_scores.mean(axis=1)

# cetak nilai akhir
print("Skor train pada ukuran data terbesar :",
    round(train_mean[-1],3))
print("Skor validasi pada ukuran data terbesar:",
    round(val_mean[-1],3))
print("Gap (indikasi overfitting jika besar)   :",
    round(train_mean[-1] - val_mean[-1],3))

# membuat grafik
plt.plot(train_sizes,train_mean,'o-',
    label='Training Score')
plt.plot(train_sizes,val_mean,'o-',
    label='Validation Score')
plt.xlabel('Jumlah Data Latih')
plt.ylabel('Akurasi')
plt.title('Learning Curve')
plt.legend()
plt.show()
```

## Pengujian kedalaman Pohon Keputusan

```python
# Daftar kedalaman yang akan diuji
depths = [1, 2, 3, 4, None]

# Setup Canvas untuk Grid Plot
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

# Loop pembuatan model dan plot
for i, depth in enumerate(depths):
    # Buat Pipeline lengkap
    model2 = Pipeline([
        ('prep', prep),
        ('pilih', SelectKBest(f_classif, k=6)), # Memilih 6 fitur terbaik
        ('model', DecisionTreeClassifier(max_depth=depth, random_state=42))
    ])

    # Hitung Learning Curve
    train_sizes, train_scores, val_scores = learning_curve(
        model2, X_train, y_train,
        cv=5, train_sizes=np.linspace(0.1, 1.0, 10),
        scoring="accuracy"
    )

    train_mean = train_scores.mean(axis=1)
    val_mean = val_scores.mean(axis=1)
    gap = train_mean[-1] - val_mean[-1]
    # Plotting ke Subplot
    ax = axes[i]
    ax.plot(train_sizes, train_mean, 'o-', color='#1f77b4', label='Training Score')
    ax.plot(train_sizes, val_mean, 'o-', color='#ff7f0e', label='Validation Score')
    title = f'max_depth = {depth}' if depth is not None else 'max_depth = None (Max)'
    ax.set_title(title)
    ax.set_xlabel('Jumlah Data Latih')
    ax.set_ylabel('Akurasi')
    ax.legend()
    ax.grid(True, linestyle='--', alpha=0.7)
    print(f"max_depth={depth}: train={train_mean[-1]:.3f}, val={val_mean[-1]:.3f}, gap={gap:.3f}")

# Hapus subplot ke-6 yang kosong
fig.delaxes(axes[5])
plt.tight_layout()
plt.show()
```

`max_depth=1` dan `max_dept=2` (Underfitting). Jika dilihat dari grafik oranye, pada nilai `max_depth=1` dan `max_dept=2` validation hampir datar. Ini menandakan bias-nya tinggi.

Analogi: Bayangkan seorang siswa yang menebak jawaban "C" untuk semua soal ujian. Saat latihan dia mendapatkan nilai 30. Kemudian saat ujian asli (validation) dia juga mendapatkan nilai 30. Jarak nilai nol, sehingga dia tidak benar-benar belajar (underfitting)

`max_depth=3` (Good fit) terlihat garis validasi (oranye) mendekati garis biru (training). Jika dilihat dari data  `max_depth=3: train=0.805, val=0.786, gap=0.003` nilai validasi tertinggi dari nilai `max_depth` yang ada dengan nilai `gap` yang kecil.

`max_depth=4` (Good fit) terlihat garis validasi (oranye) mendekati garis biru (training), akan tetapi mulai overfitting. Jika dilihat dari data  `max_depth=4: train=0.814, val=0.785, gap=0.009` nilai validasi tertinggi dari nilai `max_depth` yang ada dengan nilai `gap` yang kecil.

`max_depth=None` sudah mulai overfitting, karena gap antara garis biru dan garis oranye semakin lebar dibandingkan dengan `max_depth=3` atau `max_depth=4`
